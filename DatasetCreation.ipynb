```
import json
```

```
# Load the dataset that needs cleaning
with open("/content/New_Srilanka_law_dataset.json", "r", encoding="utf-8") as file:
    raw_data = json.load(file)
```

```
# Clean and reformat the dataset
cleaned_data = []

for entry in raw_data:
    # Extract necessary data
    if "instruction" in entry and "output" in entry:
        cleaned_entry = {
            "question": entry["instruction"].strip(),
            "answer": entry["output"].strip()
        }
        # Skip entries with empty or incomplete content
        if cleaned_entry["question"] and cleaned_entry["answer"]:
            cleaned_data.append(cleaned_entry)

# Save the cleaned dataset
output_path = "/content/Srilanka_Law_dataset.json"
with open(output_path, "w", encoding="utf-8") as file:
    json.dump(cleaned_data, file, indent=4, ensure_ascii=False)

output_path
```
