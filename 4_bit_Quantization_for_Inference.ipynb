```
!pip install bitsandbytes
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline, BitsAndBytesConfig

model_name = "Midu2712/Mistral-7B-Instruct-v0.1-finetuned_Law"

# Load model with 4-bit quantization
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True
)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    quantization_config=bnb_config,
    torch_dtype="auto"
)

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Inference pipeline with optimized settings
pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=200,
    temperature=0.7,    # Balanced randomness for coherent outputs
    top_k=50,
    top_p=0.95
)

# Prompt for testing
prompt = "What is the punishment for theft under Sri Lankan law?"
result = pipe(f"### Instruction:\n{prompt}\n\n### Response:\n")
print(result[0]['generated_text'].strip())
```
