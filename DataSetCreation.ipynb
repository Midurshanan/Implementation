```python
import pandas as pd
import json
```

```python
df1 = pd.read_csv("/content/LawDataset1.csv")
df3 = pd.read_csv("/content/LawDataset2.csv")
df2 = pd.read_csv("/content/LawDataset3.csv")
df4 = pd.read_csv("/content/LawDataset4.csv")
df5 = pd.read_csv("/content/LawDataset5.csv")
```

```python
def format_dataset(df):
    formatted_data = []

    for _, row in df.iterrows():
        user_question = row["output"].replace("Question:", "").strip()
        assistant_response = row["instruction"].strip()

        formatted_data.append({"role": "user", "content": user_question})
        formatted_data.append({"role": "assistant", "content": assistant_response})

    return formatted_data


formatted_data1 = format_dataset(df1)
formatted_data2 = format_dataset(df2)
formatted_data3 = format_dataset(df3)
formatted_data4 = format_dataset(df4)
formatted_data5 = format_dataset(df5)

# Merge both datasets
merged_data = formatted_data1 + formatted_data2 + formatted_data3 + formatted_data4 + formatted_data5
```

```python
output_path = "/content/cleaned_law_dataset.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(merged_data, f, indent=4, ensure_ascii=False)
```
